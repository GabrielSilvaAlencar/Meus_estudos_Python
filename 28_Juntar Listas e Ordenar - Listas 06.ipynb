# Juntar Listas, Ordenar e Cuidados Especiais

### 2 formas:

- lista1.extend(lista2)
- lista_nova = lista1 + lista2

Obs: o Método .append não junta listas, mas adiciona um valor no final da lista

```python
produtos = ['apple tv', 'mac', 'iphone x', 'iphone 11', 'IPad', 'apple watch', 'mac book', 'airpods']
novos_produtos = ['iphone 12', 'ioculos']

todos_produtos = produtos + novos_produtos #aqui usamos o '+' justamente para adicionar itens a lista e juntar na principal
print (todos_produtos)

# se fosse ma maneira antiga 'append' ela iria adicionar uma lista dentro da outra.
```

### Cuidado:

- [1] + [2] não é a mesma coisa que 1 + 2, então cuidado sempre com o formato dos valores na hora de fazer operações.

```python
vendas = [1000, 1500, 15000, 20000, 270, 900, 100, 1200]
vendas_iphonex = [15000]
vendas_iphone11 = [20000]

total_iphone = vendas[2] + vendas[3] #somei o total usando o indice 
print(total_iphone)

total_iphone_listas = vendas_iphone11[0] + vendas_iphonex [0] #aqui usei as variaveis, mas a soma a logica é a mesma
print(total_iphone_listas)

```

### Ordenar listas

lista.sort()

```python
produtos.sort()
print(produtos)    # aqui ele alinha por ordem alfabetica, mas eles tem regras diferentes, sempre letra maiuscula começa primeiro.
# para tratar isso é outro comando.


produtos.sort(reverse=produtos) # como o nome ja diz ele inverte as ordens.
print(produtos)
```
