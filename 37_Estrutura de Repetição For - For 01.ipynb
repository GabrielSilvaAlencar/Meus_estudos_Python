# Estrutura de Repetição: for

### Funcionamento:

```python
for i in range (5):  # para 'i' seria o nome da variavel, dentro de um ranger(5) repetir o print 5 vezes.
    print('Lira')    # o 'i' é o indice, na primeira vez ele seria o 0, na segunda o 1 e por ai vai, ate dar o ranger total
```

- Imagine que você está construindo uma automação para enviar todo dia por e-mail um resumo da produção de uma fábrica. Construa um código que exiba a quantidade produzida de cada os produto nesse "e-mail"

```python
produtos = ['coca', 'pepsi', 'guarana', 'sprite', 'fanta']
producao = [15000, 12000, 13000, 5000, 250]

tamanho = len(produtos) # para descobrir quantos produtos tenho na lista 
print(tamanho)

for i in range (5):         # para cada variavel que tem dentro do range contar ate 5
    print (' {} unidades produzidas de {}'.format(producao[i], produtos[i]))
```
